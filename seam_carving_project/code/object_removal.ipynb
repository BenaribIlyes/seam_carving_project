###  Suppression d'objet par Seam Carving

Ce notebook montre comment utiliser le Seam Carving pour **supprimer des objets spécifiques** d’une image 
en les forçant à être inclus dans les seams supprimés. (masque fourni)

Fonctionnalités :
- suppression guidée par une **carte d'énergie modifiée** (pénalisation de zones à supprimer),
- visualisation pas à pas du processus.

Permet de valider l'efficacité de Seam Carving dans des tâches de retouche ciblée.

```python
# Install missing dependencies
%pip install numba
# For displaying images in Jupyter notebooks
%matplotlib inline

# Import necessary libraries

import cv2 # type: ignore
import numpy as np
import matplotlib.pyplot as plt

# To use joblib for parallel processing
from joblib import Parallel, delayed

from class_SeamCarver import SeamCarver
```

```python
image_ref = cv2.imread('../dataset/broadway_original.webp') # Load the reference image
mask = cv2.imread('../mask/mask_broadway_original.png', cv2.IMREAD_GRAYSCALE) # Load the mask image in grayscale
```

```python
sc = SeamCarver(image_ref)

image_without_object, image_without_object_scaled_up= sc.remove_object(image_ref, mask, method='l1',orientation='vertical')

plt.figure(figsize=(12, 6))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(image_ref, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(image_without_object.astype(np.uint8), cv2.COLOR_BGR2RGB))
plt.title('Object removed')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(image_without_object_scaled_up.astype(np.uint8), cv2.COLOR_BGR2RGB))
plt.title('Final image')
plt.axis('off')

plt.tight_layout()
plt.show()

print("Original Image Shape:", image_ref.shape)
print("Object removed:", image_without_object.shape)
print("Final image:", image_without_object_scaled_up.shape)
```
